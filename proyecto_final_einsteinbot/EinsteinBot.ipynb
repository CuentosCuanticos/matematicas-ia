\[\texttt{stoi}:\text{token}\mapsto\text{entero},\qquad \texttt{itos}:\text{entero}\mapsto\text{token}.\]

```python
SPECIAL=['<pad>','<unk>','<bos>','<eos>','<user>','<assistant>']
TOKEN_RE=re.compile(r'<user>|<assistant>|<bos>|<eos>|[A-Za-zÁÉÍÓÚÜÑáéíóúüñ]+|\d+(?:[.,]\d+)?|[^\w\s]',re.UNICODE)
def tokenize(text): return TOKEN_RE.findall(text.lower())

tokens=tokenize(corpus); counts=Counter(tokens)
vocab=list(SPECIAL)
for tok,_ in counts.most_common(4000-len(SPECIAL)):
    if tok not in vocab: vocab.append(tok)
stoi={t:i for i,t in enumerate(vocab)}; itos={i:t for t,i in stoi.items()}
UNK=stoi['<unk>']; BOS=stoi['<bos>']; EOS=stoi['<eos>']

def encode(text): return [stoi.get(t,UNK) for t in tokenize(text)]
def decode(ids):
    s=' '.join(itos[int(i)] for i in ids)
    return re.sub(r'\s+([,.;:!?])',r'\1',s)

print('Tokens:',len(tokens),'Vocabulario:',len(vocab))
```

```python
class MiniGPT(nn.Module):
    def __init__(self,vocab_size,block_size,d_model=96,n_heads=4,n_layers=3,dropout=.05):
        super().__init__(); self.block_size=block_size
        self.tok=nn.Embedding(vocab_size,d_model); self.pos=nn.Embedding(block_size,d_model)
        self.blocks=nn.ModuleList([Block(d_model,n_heads,block_size,dropout) for _ in range(n_layers)])
        self.ln=nn.LayerNorm(d_model); self.head=nn.Linear(d_model,vocab_size,bias=False)
        self.apply(self._init); self.head.weight=self.tok.weight
    @staticmethod
    def _init(m):
        if isinstance(m,(nn.Linear,nn.Embedding)):
            nn.init.normal_(m.weight,mean=0.,std=.02)
            if isinstance(m,nn.Linear) and m.bias is not None: nn.init.zeros_(m.bias)
    def forward(self,idx,targets=None):
        B,T=idx.shape; pos=torch.arange(T,device=idx.device)
        x=self.tok(idx)+self.pos(pos)[None,:,:]
        for b in self.blocks: x=b(x)
        logits=self.head(self.ln(x)); loss=None
        if targets is not None: loss=F.cross_entropy(logits.reshape(-1,logits.size(-1)),targets.reshape(-1))
        return logits,loss

model=MiniGPT(len(vocab),BLOCK_SIZE).to(device)
n_params=sum(p.numel() for p in model.parameters())
print('Parámetros:',n_params)
```

# E. Entrenamiento
Minimizamos
\[\mathcal L(\theta)=-\frac1N\sum_n\log p_\theta(t_{n+1}\mid t_{\le n}),\qquad \mathrm{PPL}=e^{\mathcal L}.\]

```python
@torch.no_grad()
def estimate_loss(model,train_data,val_data,iters=15):
    model.eval(); out={}
    for name,data in [('train',train_data),('val',val_data)]:
        ls=[]
        for _ in range(iters):
            xb,yb=get_batch(data); _,loss=model(xb,yb); ls.append(loss.item())
        out[name]=float(np.mean(ls))
    model.train(); return out

def train_model(model,train_data,val_data,steps=300,lr=3e-3,eval_every=50):
    opt=torch.optim.AdamW(model.parameters(),lr=lr); hist={'step':[],'train':[],'val':[]}
    for step in range(steps+1):
        if step%eval_every==0 or step==steps:
            L=estimate_loss(model,train_data,val_data); hist['step'].append(step); hist['train'].append(L['train']); hist['val'].append(L['val']); print(step,L,'PPL',math.exp(L['val']))
        if step==steps: break
        xb,yb=get_batch(train_data); _,loss=model(xb,yb); opt.zero_grad(set_to_none=True); loss.backward(); torch.nn.utils.clip_grad_norm_(model.parameters(),1.); opt.step()
    return hist

# TODO 5: cuando hayas resuelto TODO 2 y TODO 3, ejecuta:
# history_pre=train_model(model,train_ids,val_ids,steps=300,lr=3e-3)
```
