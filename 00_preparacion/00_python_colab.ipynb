## 3. Qué necesitas saber de Python

Para seguir el curso no hace falta dominar Python.

Solo utilizaremos unas pocas ideas:

- variables;
- listas y arrays;
- operaciones matemáticas;
- funciones;
- bucles sencillos;

- representación gráfica de datos.

El objetivo de este cuaderno es comprobar que esas operaciones básicas resultan familiares antes de comenzar los laboratorios matemáticos.

```python
# Asignamos el valor 3 a la variable a.
a = 3

# Asignamos el valor 5 a la variable b.
b = 5

# Sumamos los valores almacenados en a y b.
suma = a + b

# Multiplicamos los valores almacenados en a y b.
producto = a * b

# Elevamos a al cuadrado.
potencia = a**2

# Mostramos el valor de a.
print("a =", a)

# Mostramos el valor de b.
print("b =", b)

# Mostramos el resultado de la suma.
print("a + b =", suma)

# Mostramos el resultado del producto.
print("a · b =", producto)

# Mostramos el cuadrado de a.
print("a² =", potencia)
```

```python
# Creamos una lista de Python con cuatro números.
lista = [1, 2, 3, 4]

# Convertimos la lista en un array de NumPy.
# Los arrays permiten hacer operaciones matemáticas sobre todos
# sus elementos de forma simultánea.
vector = np.array(lista)

# Mostramos en pantalla la lista original de Python.
print("Lista de Python:", lista)

# Mostramos el array de NumPy.
print("Array de NumPy:", vector)

# Multiplicamos cada componente del vector por 2.
print("2 · vector =", 2 * vector)

# Elevamos al cuadrado cada componente del vector.
print("vector² =", vector**2)
```

## 4. Funciones y bucles

En los laboratorios necesitaremos repetir operaciones y encapsular cálculos en funciones.

Una función recibe unos datos de entrada, realiza una serie de operaciones y devuelve un resultado. Los bucles permiten repetir una instrucción varias veces.

Estas dos ideas aparecerán constantemente al construir algoritmos de aprendizaje paso a paso.

```python
# Definimos una función llamada cuadrado.
# La palabra def indica a Python que estamos creando una función.
# Entre paréntesis escribimos el dato de entrada de la función.
def cuadrado(x):

    # La instrucción return devuelve el resultado de la función.
    # En este caso devolvemos x elevado al cuadrado.
    return x**2


# Creamos una lista con varios números.
numeros = [1, 2, 3, 4, 5]


# El bucle for recorre los elementos de la lista uno a uno.
# En cada iteración, la variable numero toma uno de los valores de la lista.
for numero in numeros:

    # Llamamos a la función cuadrado usando numero como dato de entrada.
    resultado = cuadrado(numero)

    # Mostramos el número original y el resultado calculado.
    print(numero, "→", resultado)
```
